Q1. Présentation de l'algorithme. Rédigez une présentation synthétique

La régression linéaire (Linear Regression) est un algorithme d'apprentissage supervisé appartenant à la famille de la régression. Son objectif est de prédire une valeur numérique continue à partir d'une ou plusieurs variables explicatives. Son principe consiste à ajuster une droite ou un hyperplan qui représente la relation entre les variables d'entrée et la variable cible. L'algorithme utilise la méthode des moindres carrés ordinaires (Ordinary Least Squares) pour trouver les coefficients qui minimisent la somme des carrés des erreurs. Le modèle commence par recevoir les données, construit une combinaison pondérée des variables, estime les coefficients, puis utilise ces coefficients pour effectuer de nouvelles prédictions.

Formule mathématique : [ \hat{y}i=\beta_0+\beta_1x{i1}+\cdots+\beta_px_{ip} ]

La fonction à minimiser est : [ J(\beta)=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2 ]

Définition des symboles :

(y_i) : valeur réelle.
(\hat{y}_i) : valeur prédite.
(x_{ij}) : valeur de la variable explicative (j) pour l'observation (i).
(\beta_0) : constante du modèle.
(\beta_j) : coefficient associé à la variable (j).
(n) : nombre d'observations.
(p) : nombre de variables explicatives.
(J(\beta)) : somme des carrés des erreurs. Paramètres appris : les coefficients (\beta_0,\beta_1,\ldots,\beta_p), estimés pendant l'apprentissage.
Hyperparamètres : la régression linéaire ordinaire n'a pas d'hyperparamètre principal de régularisation. Dans Scikit-learn, fit_intercept permet notamment de choisir si une constante est ajoutée au modèle.

Avantages :

Simple à comprendre et à interpréter.
Rapide à entraîner et à utiliser. Limites :
Elle suppose une relation linéaire entre les variables.
Elle est sensible aux valeurs extrêmes et aux relations complexes non linéaires. Point de vigilance : les coefficients indiquent des variations conditionnelles entre les variables, mais ne prouvent pas une relation de causalité.

2. Nature des données et prérequis techniques
La régression linéaire nécessite :
- Des données structurées sous forme de tableau.
- Des variables explicatives numériques ou correctement encodées.
- Une variable cible numérique continue.
- Une relation approximativement linéaire entre les variables explicatives et la variable cible.
- Des données suffisamment propres pour permettre un apprentissage fiable.
L'algorithme peut être utilisé avec Python grâce à la bibliothèque Scikit-learn, notamment avec la fonction LinearRegression().

**3. Cas où l'algorithme est peu approprié**
La régression linéaire est peu appropriée lorsque les relations entre les variables sont fortement non linéaires.

Par exemple, si l'évolution des crédits bancaires dépend de relations économiques complexes qui ne peuvent pas être représentées par une droite, le modèle risque de produire des prédictions imprécises.

Elle est également sensible aux valeurs atypiques (outliers).

**4. Métriques d'évaluation**
Pour évaluer les performances du modèle, nous pouvons utiliser :
Métrique	Signification
MAE	Erreur absolue moyenne entre les valeurs réelles et prédites.
MSE	Moyenne des erreurs au carré.
RMSE	Racine carrée de la moyenne des erreurs au carré.
R²	Mesure la part de la variation de la cible expliquée par le modèle.

Interprétation : une MAE et une RMSE faibles indiquent généralement de meilleures prédictions. Un coefficient R² proche de 1 indique que le modèle explique une grande partie de la variation de la variable cible.

***2. Q7.2 – Entraînement du modèle de régression linéaire***

Nous allons utiliser le temps comme variable explicative pour estimer l'évolution du crédit bancaire.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Créer une variable temporelle
df_credit = df_credit.reset_index(drop=True)
df_credit["Temps"] = np.arange(len(df_credit))

# Définir X et y
X = df_credit[["Temps"]]
y = df_credit["Credit_bancaire"]

# Séparation chronologique : 80% entraînement, 20% test
limite = int(len(df_credit) * 0.8)

X_train = X.iloc[:limite]
X_test = X.iloc[limite:]

y_train = y.iloc[:limite]
y_test = y.iloc[limite:]

# Entraîner le modèle
modele = LinearRegression()
modele.fit(X_train, y_train)

# Prédictions
y_pred = modele.predict(X_test)

# Évaluation
print("MAE :", mean_absolute_error(y_test, y_pred))
print("RMSE :", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R² :", r2_score(y_test, y_pred))
```

***Interprétation des métriques d'évaluation***


L'évaluation du modèle de régression linéaire donne les résultats suivants :
- MAE = 66 104,93 : l'erreur absolue moyenne entre les valeurs réelles et les prédictions est d'environ 66 105 MDH.
- RMSE = 67 649,22 : cette valeur montre également un écart important entre les prédictions et les valeurs réelles.
- R² = -0,485 : le coefficient de détermination négatif indique que le modèle prédit moins bien que le modèle de référence qui utilise la moyenne des valeurs d'entraînement.
Conclusion : les métriques confirment que la régression linéaire simple présente des performances insuffisantes pour prévoir le crédit bancaire. Le modèle ne capture pas correctement les variations de la série temporelle. Il serait donc intéressant de tester d'autres algorithmes afin d'améliorer la qualité des prédictions.

***3. Visualisation des prédictions***

```python
plt.figure(figsize=(12, 5))

plt.plot(df_credit["Date"], y,
         label="Valeurs réelles")

plt.plot(df_credit["Date"].iloc[limite:],
         y_pred,
         label="Prédictions du modèle",
         linestyle="--")

plt.title("Régression linéaire - Crédit bancaire au Maroc")
plt.xlabel("Date")
plt.ylabel("Crédit bancaire (MDH)")
plt.legend()
plt.grid()
plt.show()
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## Annexes — implémentations manuelles des trois algorithmes imposés

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
